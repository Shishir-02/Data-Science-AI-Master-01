```python
# Even Odd Finder

def even_odd(num)-> None:
    if num % 2 == 0:
        print(num, "is an even number")
    else:
        print(num,"is an odd number")
```

```python
even_odd(101)
```

```python
# Factorial finder

def find_factorial(num)->int:
    fact = 1
    for i in range(1,num + 1):
        fact = fact*i
    return fact
```

```python
find_factorial(5)
```

```python
# Sum of natural nubers till the provided number

def sum_of_natural(num) ->int:
    sum = 0
    for i in range (1,num+1):
        sum = sum+i
    return sum
```

```python
sum_of_natural(6)
```

```python
# Function with default value arguments

def functionOne(name, age = 30)->None:
    print("User Name: ", name)
    print ("Age: ", age)
```

```python
functionOne("Shahadat") # With out passing age value returns default value
```

```python
functionOne("Shahadat", 25) # Now with passing age value returns updated age value
```

### Lambda Functions

Lambda function diclaration format -> lambda(key word) argument : expression

```python
x = lambda arg : arg+10
```

```python
print (x(5))
```

```python
# Let's create a lambda function to check even and odd number
x = lambda arg : "Even" if arg % 2 == 0 else "Odd"
print(x(100))
```

```python
# Let's create a lambda function to add tow integers
x = lambda arg1,arg2 : arg1 + arg2
print(x(2,3))
```
