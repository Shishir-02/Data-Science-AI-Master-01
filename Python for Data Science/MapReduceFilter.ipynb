```python
import math

def areaCalculator(r):
    return math.pi*(r**2)

rediousList = [1,2,3,4,5,6]
areas = []

for r in rediousList:
    a = areaCalculator(r)
    areas.append(a)

areas
```

```python
# Let's use Map function to avoid For - Loop
map(areaCalculator,rediousList)
list(map(areaCalculator,rediousList))
```

Converting Celcius scale to Farheneit

```python
temps = [("Dhaka", 28), ("Mumbai", 35), ("Berlin", 12), ("Tokyo", 25), ("Hong Kong", 24), ("Sydney", 30)]
```

```python
type(temps)
```

```python
# F = 9/5C + 32
cel_to_ferhen = lambda data: (data[0], (9/5*data[1])+32)
map(cel_to_ferhen,temps)
```

```python
list(map(cel_to_ferhen,temps))
```
