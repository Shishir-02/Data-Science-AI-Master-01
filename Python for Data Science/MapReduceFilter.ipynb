## Map Function

The "Map" function is used to apply a specific function type operation on a dataset. Simply f(x) type operation, to every item in an iterable dataset (like a list, tuple, or dictionary), which returns a new list of data set, with applied f(x) on every elements.

Syntax: map(f(x),data)

```python
# Traditional Loop Approach
import math
def areaCalculator(r)->None: 
    return math.pi*(r**2)

rediousList = [1,2,3,4,5,6]
areas = []
for r in rediousList:
    a = areaCalculator(r)
    areas.append(a)
areas
```

```python
# Let's use Map function to avoid For - Loop
map(areaCalculator,rediousList)
list(map(areaCalculator,rediousList))
```

Converting Celcius scale to Farheneit

## Filter Function

Filter function is a builtin function that allows us to process and iterate, extract items based on some conditions. It simply filters out the necessary data. Using filter function we can also get rid of "Null Values".

Syntext: filter(f(x),data)

```python
import statistics
data = [1,2,5,4,6,8,2,5,4,3]
average = statistics.mean(data)
print(average)
```

```python
# Lets apply filter function
filteredAverage = list(filter(lambda x: x>average,data)) # Filtering values which are greater than average.
filteredAverage
```


Let's get rid of Null/Nan values

```python
names = ["Shahadat", "Neha", "Sumu", "", 0, 0.0, "", "Kabir"]
nullfilteredNames = list(filter(None, names))
nullfilteredNames
```

```python
numbers = [12,52,60,1,8,0,0.0,77,43,"",99]
nullFilteredNumbers = list(filter(None,numbers))
nullFilteredNumbers
```

## Reduce Function

The pythons's default "Reduce" function has been depricated ! Therefore, we use "Functools Reduce" function. Let's try this in practice. Reduce function is used to perform "Rolling Computation" across the dataset.

Syntax: reduce(f(x),data)

```python
from functools import reduce

# Multiply of all the items in a list
data = [1,2,3,4,5,3]
multiplier = lambda x,y : x*y
reduce(multiplier,data)
```

```python
# Addition of all the items in a list
data2 = [5,0,3,1,7,10]
addResult = lambda x,y : x+y
reduce(addResult,data2)
```
