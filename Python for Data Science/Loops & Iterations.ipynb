# For loop

```python
# Iterate over a list of integers using for loop
list = [10,20,15,25,60,34,40,50,55,61]
for i in list:
    print(i)
```

```python
# Iterate over a string using for loop
string = "Bangladesh"
for i in string:
    print(i)
```

```python
# Iterate over a dictionary using for loop
student_info = {1:"[Name: John, Age: 20]", 2:"[Name: Alice, Age: 22]", 3:"[Name: Bob, Age: 19]", 4:"[Name: David, Age: 21]"}
for i in student_info:
    print(i, ":", student_info[i])
```

```python
# Iterate over a dictionary using for loop
student_info = {1:"[Name: John, Age: 20]", 2:"[Name: Alice, Age: 22]", 3:"[Name: Bob, Age: 19]", 4:"[Name: David, Age: 21]"}
for i in student_info:
    print(student_info[i])
```

```python
# Iterate over a dictionary using for loop
student_info = {1:"[Name: John, Age: 20]", 2:"[Name: Alice, Age: 22]", 3:"[Name: Bob, Age: 19]", 4:"[Name: David, Age: 21]"}
for i,j in student_info.items():
    print(j)
```

## Comprehensions 
### are used instead of using for-loop to speed up the execution process. It has less time complexity in comparison with traditional for-loop

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia", "Toyota", "BMW"]
list_x = ["Countries", "Bangladesh", "Franch", "Canada", "Pakistan", "India", "Nepal"]
list_2 = []

for i in list_1:
    list_x.append(i)
print (list_x,"\n")

for i in list_1:
    list_2.append(i)
print(list_2)
```

```python
list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia", "Toyota", "BMW"]
list_2 = []

for i in list_1:
    list_2.append(len(i))
print("With Out comprehension: ",list_2)


# Same things with List Comprehension
list_2 = [len(i) for i in list_1]
print("With comprehension: ", list_2)
```

```python
# Let's append list and each element's length with out and with using list comprehension

list_1 = ["Automobiles", "Honda", "Benz", "Maruti", "Kia", "Toyota", "BMW"]
dictionary = {}

# with out using list comprehension
for i in list_1:
    dictionary[i] = len(i)
print("With out List Comprehension: ",dictionary,"\n")

# with list comprehension
dictionary = {i:len(i) for i in list_1}
print("With using Comprehension: ", dictionary)
```
