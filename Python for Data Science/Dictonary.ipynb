Data in dictionary is stored as a Key:value pair. Where,
Key should not be mutable/changable but values can be of any type.

```python
Dict = {"name": "John", "age": 30, "city": "New York"}
```

```python
type(Dict)
```

```python
Dict2 = {"India": "INR", "USA": "USD", "Hong Kong": "HKD", "Japan": "JPY"}
```

```python
type(Dict2)
```

```python
len(Dict2)
```

```python
# Accessing Values in a Dictionary
Dict2["India"]  # Output: 'INR'
```

```python
# Replacing Values in a Dictionary
Dict2["India"] = "Indian Rupee"  # Replaces the value for the key "India"
Dict2["India"]
```

```python
Dict2["Japan"] = "YEN"  # Adding a new key-value pair to the dictionary
```

```python
Dict2
```

```python
# Deleting a Key-Value Pair from a Dictionary
del Dict2["USA"]  # Deletes the key-value pair for "USA"
Dict2
```

```python
len(Dict2)
```

```python
# Sorting a dictionary
sorted(Dict2)
```

```python
# values() mehtod returns a view object that displays a list of all the values in the dictionary.
Dict2.values()
```

```python
# keys() method returns a view object that displays a list of all the keys in the dictionary.
Dict2.keys()
```

```python
# get() method returns the value for the specified key if the key is in dictionary.
Dict2.get("India")
```

```python
# update() method updates the dictionary with the elements from another dictionary object or from an iterable of key-value pairs.
Dict2.update({"India": "Rs"})
```
